<h2> Code to inspect evaluation dataframes </h2>

Mean explanation sizes for cf-gnnexplainer can appear smaller than results for bruteforce.
Functions in this notebook investigate if this is cause by its lower fideltity

```python
import pandas as pd


def compare_bf_nan(dst, comparison='bf'):
    with open(f'../../results/{dst}sparsecf.pkl', "rb") as f:
        cfgnnx = pd.read_pickle(f)

    with open(f'../../results/{dst}sparse{comparison}.pkl', "rb") as f:
        bf = pd.read_pickle(f)

    a = cfgnnx.loc[:, ['node', 'label', 'distance']]
    a['distance_bf'] = bf['distance']

    a.loc[(a.loc[:, 'distance'] < a.loc[:, 'distance_bf']), :]
    no_na = a.dropna()
    no_na.distance.mean()
    print(f'average distance cf-gnnexplainer: {a.distance.mean():.3f}')
    print(f'average distance normal: {a.distance_bf.mean():.3f}')
    print(f'average distance nan dropped: {no_na.distance_bf.mean():.3f}')

def compare_bf_nan_original(dst, comparison='bf'):
    with open(f'../../results/{dst}original.pkl', "rb") as f:
        cfgnnx = pd.read_pickle(f)

    with open(f'../../results/{dst}{comparison}.pkl', "rb") as f:
        bf = pd.read_pickle(f)

    a = cfgnnx.loc[:, ['node', 'label', 'distance']]
    a['distance_bf'] = bf['distance']

    a.loc[(a.loc[:, 'distance'] < a.loc[:, 'distance_bf']), :]
    no_na = a.dropna()
    no_na.distance.mean()
    print(f'average distance cf-gnnexplainer: {a.distance.mean():.3f}')
    print(f'average distance normal: {a.distance_bf.mean():.3f}')
    print(f'average distance nan dropped: {no_na.distance_bf.mean():.3f}')

```

```python
print('bruteforce syn1, syn2 sparse gradients')
compare_bf_nan('syn1')
print('')
compare_bf_nan('syn2')
print('')
print('greedy syn1, syn2 sparse gradients')
compare_bf_nan('syn1', 'greedy')
print('')
compare_bf_nan('syn2', 'greedy')
print('')

```

```python
print('bruteforce syn1, syn2 original gradients')
compare_bf_nan_original('syn1')
print('')
compare_bf_nan_original('syn2')
print('')
print('greedy syn1, syn2 original gradients')
compare_bf_nan_original('syn1', 'greedy')
print('')
compare_bf_nan_original('syn2', 'greedy')
```
